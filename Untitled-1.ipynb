User module


```python
class User:
    def __init__(self, username, email, __password, phone_number, Address, Order_history):
        self.username = username
        self.email = email
        self.__password = __password
        self.phone_number = phone_number
        self.Address = Address
        self.Order_history = Order_history
    def register(self):
        print(f"User {self.username} ({self.email}) ({self.__password}) ({self.phone_number}) ({self.Address}) registered successfully.")
    def login(self):
        print(f"User {self.username} ({self.__password}) logged in successfully.")
    def logout(self):
        print(f"User {self.username} logged out successfully.")
    def view_profile(self):
        print(f"User {self.username} ({self.email}) ({self.__password}) ({self.phone_number}) ({self.Address}) profile viewed successfully.")
    def update_profile(self, username, email, __password, phone_number, Address):
        self.username = username
        self.email = email
        self.__password = __password
        self.phone_number = phone_number
        self.Address = Address
        print(f"User {self.username} ({self.email}) ({self.__password}) ({self.phone_number}) ({self.Address}) profile updated successfully.")
    def change_password(self, __password):
        self.__password = __password
        print(f"User {self.username} ({self.__password}) password changed successfully.")
    def view_order_history(self):
        print(f"User {self.username} ({self.Order_history}) order history viewed successfully.")
```

Product Management


```python
class Product:
    def __init__(self, product_name, product_description, product_price, product_quantity):
        self.product_name = product_name
        self.product_description = product_description
        self.product_price = product_price
        self.product_quantity = product_quantity
    def add_product(self):
        print(f"Product {self.product_name} ({self.product_description}) ({self.product_price}) ({self.product_quantity}) added successfully.")
    def update_product(self, product_name, product_description, product_price, product_quantity):
        self.product_name = product_name
        self.product_description = product_description
        self.product_price = product_price
        self.product_quantity = product_quantity
        print(f"Product {self.product_name} ({self.product_description}) ({self.product_price}) ({self.product_quantity}) updated successfully.")
    def view_product(self):
        print(f"Product {self.product_name} ({self.product_description}) ({self.product_price}) ({self.product_quantity}) viewed successfully.")
    def change_price(self, product_price):
        self.product_price = product_price
        print(f"Product {self.product_name} ({self.product_price}) price changed successfully.")
```

Shopping cart

```python
class Cart:
    def __init__(self, cart_items, cart_total):
        self.cart_items = cart_items
        self.cart_total = cart_total
    def add_to_cart(self, product_name, product_price):
        self.cart_items.append(product_name)
        self.cart_total += product_price
        print(f"Product {product_name} ({product_price}) added to cart successfully.")
    def remove_from_cart(self, product_name, product_price):
        if product_name in self.cart_items:
            self.cart_items.remove(product_name)
            self.cart_total -= product_price
            print(f"Product {product_name} ({product_price}) removed from cart successfully.")
        else:
            print(f"Product {product_name} not found in cart.")
    def view_cart(self):
        print(f"Cart items: {self.cart_items} (Total: {self.cart_total}) viewed successfully.")
    def increase_quantity(self, product_name, product_price):
        if product_name in self.cart_items:
            self.cart_total += product_price
            print(f"Product {product_name} ({product_price}) quantity increased in cart successfully.")
        else:
            print(f"Product {product_name} not found in cart.")
    def decrease_quantity(self, product_name, product_price):
        if product_name in self.cart_items:
            self.cart_total -= product_price
            print(f"Product {product_name} ({product_price}) quantity decreased in cart successfully.")
        else:
            print(f"Product {product_name} not found in cart.") 
    def clear_cart(self):
        self.cart_items = []
        self.cart_total = 0
        print(f"Cart cleared successfully.")
    
```

Payment system

```python
from abc import ABC, abstractmethod
def pay(amount):
    @abstractmethod
    def process_payment(self):
        pass
    def UPI(self):
        print(f"UPI payment of {amount} processed successfully.")
    def credit_card(self):
        print(f"Credit card payment of {amount} processed successfully.")
    def debit_card(self):
        print(f"Debit card payment of {amount} processed successfully.")
    def net_banking(self):
        print(f"Net banking payment of {amount} processed successfully.")
    def ccavenue(self):
        print(f"CC Avenue payment of {amount} processed successfully.")
```

Order management

```python
class Order:
    def __init__(self, order_id, order_date, order_status, order_total):
        self.order_id = order_id
        self.order_date = order_date
        self.order_status = order_status
        self.order_total = order_total
    def place_order(self):
        print(f"Order {self.order_id} ({self.order_date}) ({self.order_status}) ({self.order_total}) placed successfully.")
    def cancel_order(self):
        print(f"Order {self.order_id} ({self.order_date}) ({self.order_status}) ({self.order_total}) canceled successfully.")
    def track_order(self):
        print(f"Order {self.order_id} ({self.order_date}) ({self.order_status}) ({self.order_total}) tracked successfully.")
    def view_order_details(self):
        print(f"Order {self.order_id} ({self.order_date}) ({self.order_status}) ({self.order_total}) details viewed successfully.")
    def update_order_status(self, order_status):
        self.order_status = order_status
        print(f"Order {self.order_id} ({self.order_date}) ({self.order_status}) ({self.order_total}) status updated successfully.")
```
